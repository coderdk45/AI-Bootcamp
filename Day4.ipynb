```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("image1.jpeg")


# OpenCV reads images as BGR
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
print(img.shape)

img_resize = cv2.resize(img, (400, 256))

img_resize_rgb = cv2.cvtColor(img_resize, cv2.COLOR_BGR2RGB)

plt.imshow(img_resize_rgb)
plt.axis("off")
plt.show()


```

```python
img = cv2.imread("image1.jpeg")

# Flip image horizontally
img_flip = cv2.flip(img, 1)

# Convert BGR to RGB
img_flip_rgb = cv2.cvtColor(img_flip, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(img_flip_rgb)
plt.axis("off")
plt.show()
```
